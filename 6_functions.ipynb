```python
# multyplying factorials

import math

def mult_factorials(*numbers):
    mult = 1
    try:
        for number in numbers:
            f = math.factorial(number)
            mult *= f
    except TypeError:
        return "Not numbers in the arguments"

    return mult

mult_factorials(5,"hola")
    
```

### **2. Prefix notation calculator**

```python
# Evaluate prefix expressions.

import operator

def calc(prefix: str):
    parts = prefix.split()
    op = parts[0]
    n1 = int(parts[1])
    n2 = int(parts[2])

    operations = {"+": operator.add, "-": operator.sub,
                  "*": operator.mul, "/": operator.truediv,
                  "%": operator.mod, "**": operator.pow}

    return operations[op](n1,n2)

calc("- 1 2")
```

```python
# expanding calculator
import operator
from functools import reduce

def extended_calc(prefix: str):
    parts = prefix.split()
    op = parts[0]

    operations = {"+": operator.add, "-": operator.sub,
                  "*": operator.mul, "/": operator.truediv,
                  "%": operator.mod, "**": operator.pow}
    
    numbers = tuple(int(parts[i]) for i in range(1,len(parts)))
    #print(numbers)

    return reduce(operations[op], numbers)

extended_calc("/ 100 5 5")
```

```python
# mapping function

def apply_to_each(function, iterable):
    solution = []
    for it in iterable:
        solution.append(function(it))

    return solution

apply_to_each(str.upper, ["hola", "python", "mundo"])
apply_to_each(abs, [-1, -5, 10])

```

```python
def transform_lines(function, input_file, output_file):
    with open(input_file, "r", encoding="utf-8") as i, \
         open(output_file, "w", encoding="utf-8") as o:
        
        for line in i:
            o.write(function(line))

def add_arrow(line):
      return "-> " + line

transform_lines(str.upper, "shell.txt", "shell_upper.txt")
transform_lines(add_arrow, "shell.txt", "shell_arrows.txt")
```

### **3. Password generator**

```python
# Generate random passwords.

from random import choice

def create_password_generator(s: str):
    def size(sz: int):
        password = ""
        for i in range(sz):
            password += choice(s)
        return password

    return size

alpha_password = create_password_generator('abcdef')

print(alpha_password(10))


```

```python
# password checker
import string

def create_password_checker(min_uppercase: int,min_lowercase: int,
                            min_punctuation: int,min_digits: int):
    def potencial_password (passwd: str):
        punctuation = string.punctuation
        uppercase = string.ascii_uppercase
        lowercase = string.ascii_lowercase
        digits = string.digits
        p, u, l, d= [0,0,0,0]

        for c in passwd:
            if c in punctuation:
                p += 1
            elif c in uppercase:
                u += 1
            elif c in lowercase:
                l += 1
            elif c in digits:
                d += 1
            else:
                continue
        return (
        d >= min_digits
        and p >= min_punctuation
        and u >= min_uppercase
        and l >= min_lowercase
        )
    return potencial_password

checker = create_password_checker(
    min_uppercase=2,
    min_lowercase=3,
    min_punctuation=1,
    min_digits=2
)

print(checker("ABabc!12"))
print(checker("Aabc!12"))

    
```

```python
def doboth(f1: function, f2: function):
    def g(x):
        return f2(f1(x))
    return g

def double(x):
      return x * 2

def add_three(x):
      return x + 3

g = doboth(double, add_three)
print(g(10))
```
