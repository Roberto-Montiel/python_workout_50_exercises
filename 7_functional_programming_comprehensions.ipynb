### **1. Join numbers**

```python
# Join numbers into text.

def join_numbers(number: int):
    return ",".join([str(s) for s in number])

join_numbers(range(15))
```

```python
def join_numbers(number: int):
    return ",".join([str(s) for s in number if s <= 10])

join_numbers(range(15))
```

```python
#summing hexadecimal numbers

def summing_hexadecimal(hex_str):
    return sum(int(x,16) for x in hex_str)
```

```python
#reversing word orders

def reversing_order(filename: str):
    with open(filename,"r",encoding="utf-8") as f:
        return [" ".join(s[::-1]) for s in [l.split() for l in f]]
           
```

```python
# Sum numbers in text.

def sum_numbers(some_string: str):
    return sum(int(w)
            for w in some_string.split()
            if w.isnumeric() or (w.startswith("-") and w[1:].isnumeric()))

user_input = input("Enter integers: ")
print(sum_numbers(user_input))

#sum_numbers("10 abc 20 de44 30 55fg 40")
```

```python
def filtering_lines_file(filename: str):
    vowels = "aeiou"
    # we will use a list comprehension considering not too large files
    with open(filename, "r", encoding="utf-8") as file:
        return [
            line
            for line in file
            if len(line.rstrip("\n")) > 20
            and any(char.lower() in vowels for char in line)
        ]


text = filtering_lines_file("decrypt.txt")

for line in text:
    print(line, end="")
```

```python
#changing area code of phone numbers starting 0-5

def changing_prefix(phone_numbers: list):
    allowed_digits = "012345"
    return [
        f"{int(parts[0]) + 1:03d}-{parts[1]}-{parts[2]}"
        if parts[1][0] in allowed_digits
        else "-".join(parts)
        for parts in (p.strip().split("-") for p in phone_numbers)
      ]

changing_prefix(['123-456-7890', '123-333-4444', '123-777-8888'])
```

### **3. Flatten list**

```python
# Flatten nested lists.

def flatten(l: list) -> list:
    return [e for x in l for e in x]

flatten([[1,2], [3,4]])
```

```python
# flatting odd integers (including strings who can be converted to odd integers)

def flatten_odd_integers(l: list) -> list:
    def to_int(e):
        try:
            return int(e)
        except ValueError:
            return None
        
    return [n 
            for x in l 
            for e in x 
            for n in [to_int(e)]
            if n is not None and n % 2 ==1
    ]

flatten_odd_integers([[1, 2, "3"], ["4", "abc", 5]])
```

### **4 Pig latin translation of a file**

```python
# Translate file words.

def pig_latin(word: str):
    vowels = "aeiou"

    if word[0] in vowels:
        return word + "way"
    else:
        first_consonant = word[0]
        word = word[1:]
        return word + first_consonant + "ay"
    
def pig_latin_translation(filename: str):
    with open(filename, "r", encoding="utf-8") as f:
        return " ".join(pig_latin(one_word) 
                        for line in f 
                        for one_word in line.split())
    

pig_latin_translation("ejercicio_texto.txt")
```

### **6. Transform values**

```python
# Transform dictionary values.

def transform_values(function, d: dict) -> dict: 
    return {key: function(value) for key, value in d.items() }

d = {'a':1, 'b':2, 'c':3}
transform_values(lambda x: x*x, d)
```

```python
# extending transforming values

def transform_values_extended(f1,f2, d: dict) -> dict:
    return {key: f1(value) for key,value in d.items() if f2(key,value)}

d = {"a": 1,"b": 2,"c": 3,"d": 4}
transform_values_extended(lambda x: x * x,
                          lambda key, value: value % 2 == 0,
                          d
)
```

### **7. Almost supervocalic words**

```python
# Find supervocalic words.

def get_sv(filename: str) -> set:
    vowels = "aeiou"
    with open(filename,"r",encoding="utf-8") as f:
        return {word
                for line in f 
                if all(vowel in (word:= line.strip().lower()) for vowel in vowels)
                }

# analog solution
# def get_sv(filename):
#     vowels = {'a', 'e', 'i', 'o', 'u'}     
#     return {word.strip()        
#         for word in open(filename)         
#         if vowels < set(word.lower())}


get_sv("words.txt")
```

```python
def reading_user_preferences(filename: str) -> dict:
    with open(filename, "r", encoding="utf-8") as f:
        return {key: value 
                for line in f
                for key, value in [line.strip().split("=",1)]}
    
reading_user_preferences("config.txt")
```

```python
def reading_user_preferences_filtered(filename: str) -> dict:
    with open(filename, "r", encoding="utf-8") as f:
        return {key: int(value) 
                for line in f
                for key, value in [line.strip().split("=",1)]
                if value.isdigit()}
    
reading_user_preferences_filtered("config.txt")
```

```python
import json

def largest_cities(filename: str):
    with open(filename,"r",encoding="utf-8") as f:
        cities = json.load(f)
        return {city["city"]: city["population"] 
                for city in cities}
    
print(largest_cities("cities.json"))
len(largest_cities("cities.json"))
# obtaining less that 1000 cities (925)
# there can be cities with same name in different states
# and a dict cannot have duplicated keys
```

### **8b. Gematria part2**

```python
# Find gematria matches.

from string import ascii_lowercase

def creating_gematria_dict():
    return {letter: number
            for number, letter in enumerate(ascii_lowercase,1)
            }

GEMATRIA = creating_gematria_dict()

def gematria_for(one_word):
    return sum(GEMATRIA.get(letter,0) for letter in one_word)

def gematria_equal_words(one_word):
    score = gematria_for(one_word)
    with open("words.txt", "r", encoding="utf-8") as f: 
        return [line.strip()
                for line in f  
                if gematria_for(line.strip()) == score]


gematria_equal_words("casa")
```

```python
books = [
    (("George", "Orwell"), "1984", 9.99),
    (("Jane", "Austen"), "Pride and Prejudice", 7.50),
    (("Harper", "Lee"), "To Kill a Mockingbird", 8.75),
    (("F. Scott", "Fitzgerald"), "The Great Gatsby", 10.25),
    (("Mary", "Shelley"), "Frankenstein", 6.99),
]

def from_list_to_dict_books(books: list) -> dict:
    return {
        title: {
            "first_name": first_name,
            "last_name": last_name,
            "price": price,
        }
        for (first_name, last_name), title, price in books
    }

from_list_to_dict_books(books)
```
