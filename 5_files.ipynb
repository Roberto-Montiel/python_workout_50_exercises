```python
# summing integers in a file

def summing_integers(filename: str) -> int:
    total = 0

    with open(filename,"r") as f:
        for line in f:
            parts = line.strip().split()
            for word in parts:
                if word.isdigit():
                    total += int(word)

    return total

print(summing_integers("chapter5_12.txt"))
```

```python
# multiplying nd summing columns exercise

# helper function, is it the word an int or float number?

def int_or_float(word: str):
    try:
        return int(word)
    except ValueError:
        try:
            return float(word)
        except ValueError:
            return False
        
# diferencia is not y != 
def multiplying_summing_columns(filename: str):
    total = 0

    with open(filename,"r") as f:
        for line in f:
            parts = line.strip().split()
            if len(parts) != 2:
                continue

            a = int_or_float(parts[0])
            b = int_or_float(parts[1])

            if a is not None and b is not None:
                total += a * b
    
    return total

print(multiplying_summing_columns("chapter5_13.txt"))
```

```python
#counting vowels

def counting_vowels_file(filename: str) -> int:
    vowels = "aeiou"
    counting_vowels = {}

    with open(filename, "r") as f:
        for line in f:
            for letter in line.strip().lower():
                if letter in vowels:
                    counting_vowels[letter] = counting_vowels.get(letter, 0) + 1

    return dict(sorted(counting_vowels.items()))

print(counting_vowels_file("chapter5_14.txt"))
```

### **2. /etc/passwd to dict**

```python
# we will work based on passwd.txt file 
# getting user + user id like a dict 

def passwd_to_dict(filename: str) -> dict:
    users = {}

    with open(filename, "r") as f:
        for line in f:
            line = line.strip()
            if not line or line.startswith("#"):
                continue
            parts = line.split(":")
            users[parts[0]] = int(parts[2])

    return users

print(passwd_to_dict("passwd.txt"))
```

```python
#listing users login shells

def login_shells(filename: str) -> dict:
    login_shells_dict = {}

    with open(filename,"r") as f:
        for line in f:
            line = line.strip()
            if not line or line.startswith("#"):
                continue
            parts = line.split(":")
            user = parts[0]
            shell = parts[-1]
            login_shells_dict.setdefault(shell,[]).append(user)

    return login_shells_dict

print(login_shells("passwd.txt"))

```

```python
from collections import defaultdict

def get_factors(n):
    factors = set()
    for i in range(1, int(n**0.5) + 1):
        if n % i == 0:
            factors.add(i)
            factors.add(n // i)
    return factors

numbers = input("Type in integers separated by spaces: ")
numbers_list = [int(x) for x in numbers.split()]

factors_map = defaultdict(list)

for num in numbers_list:
    for factor in get_factors(num): 
        factors_map[factor].append(num)

print(dict(factors_map))
```

```python
# getting user id home directory and shell in a nested dict with user as the key

def more_info(filename: str) -> dict:
    users_info = {}

    with open(filename,"r") as f:
        for line in f:
            line = line.strip()

            if not line or line.startswith("#"):
                continue
            parts = line.split(":")
            user = parts[0]
            user_id = parts[2]
            home_directory = parts[5]
            shell = parts[6]

            users_info[user] = {"user_id": user_id, "home_directory": home_directory, "shell": shell}

    return users_info

print(more_info("passwd.txt"))
```

### **3.Counting words**

```python
def wordcount(filename: str):
    counting_characters = 0
    counting_words = 0
    counting_lines = 0
    counting_unique_words = set()

    with open(filename,"r") as f:
        for line in f:
            counting_lines += 1
            counting_characters += len(line) #including \n
            words = line.strip(".,!?;:\"'").split()
            counting_words += len(words)
            counting_unique_words.update(words)
    

    return {
        "characters": counting_characters,
        "lines": counting_lines,
        "words": counting_words,
        "unique_words": len(counting_unique_words),
    }
        
wordcount("wcfile.txt")
```

```python
#countinng words that appear in a specific file given by the user

from collections import Counter

def file_words_counting():
    solution = Counter()
    data = input("Please type in the filename and the " \
    "words you wanna search for separated by spaces: ")

    parts = data.split()
    if len(parts) < 2:
        return "You must type in a filename and one word at least"
    
    filename = parts[0]
    searched_words = set(parts[1:])
    try:
        with open(filename,"r") as f:
            for line in f:
                raw_words = line.split()
                clean_words = [w.strip(".,!?;:\"'") for w in raw_words]

                accepted_words = [w for w in clean_words if w in searched_words]
                solution.update(accepted_words)
    except FileNotFoundError:
        return "The file doesn´t exist, try it one more time"

    return dict(solution)

print(file_words_counting())
```

```python
#getting the five most commmon letters from all files in a specified path

import os
from collections import Counter

def five_most_common_letters(directory: str):

    counting = Counter()
    try:
        files = os.listdir(directory)

        for name in files:
            full_path = os.path.join(directory, name)
            if os.path.isfile(full_path):
                try:
                    with open(full_path,"r") as f:
                        for line in f:
                            line = line.replace(" ","").strip()
                            for char in line:
                                if char.lower().isalpha():
                                    counting.update(char.lower())
                except (UnicodeDecodeError, PermissionError):
                    continue  
    except FileNotFoundError:
        return "Not found directory"

    return counting.most_common(5)

print(five_most_common_letters(r"C:\Users\Usuario\Desktop\PROGRAMMING\GENERAL PYTHON\python_workout_50_exercises"))


```

### **4. Finding longest word**

```python
import os

def find_longest_word(filename: str):
    longest_word = ""

    with open(filename, "r", encoding="utf-8") as f:
        for line in f:
            raw_words = line.split()
            clean_words = [w.strip(".,!?;:\"'") for w in raw_words]
            clean_words = [w for w in clean_words if w]

            if not clean_words:
                continue

            current_longest = max(clean_words, key=len)
            if len(current_longest) > len(longest_word):
                longest_word = current_longest

    return longest_word

find_longest_word(r"books\pg28860.txt")

def find_all_longest_words(directory: str):
    return {
        file_name: find_longest_word(os.path.join(directory,file_name))
        for file_name in os.listdir(directory)
        if os.path.isfile(os.path.join(directory,file_name))
        and file_name.endswith(".txt")
    }

find_all_longest_words(r"C:\Users\Usuario\Desktop\PROGRAMMING\GENERAL PYTHON\python_workout_50_exercises\books")


```

```python
# MD5 is used here to generate a fingerprint of each file's content.
# This allows us to compare files, detect duplicates, or check if a file has changed.
# Note: MD5 is not suitable for security-critical use cases such as password storage.
# md5 works with bytes

import hashlib
import os

def md5_hashes(directory: str):
    files = os.listdir(directory)

    for file in files:
        full_path = os.path.join(directory,file)

        if os.path.isfile(full_path):
            md5_hash = hashlib.md5()  # New MD5 calculator for this file only
            
            with open(full_path, "rb") as f:
                # iter turns something iterable in an iterator, iter(callable/iterable, sentinel)
                # then we will read 
                for chunk in iter(lambda: f.read(8192), b""): 
                    md5_hash.update(chunk)

            file_hash = md5_hash.hexdigest() # to get the hash as a hexadecimal number 

            print(f"{file}: {file_hash}")

            
md5_hashes(".")
```

```python
from collections import Counter

def how_many_requests(filename: str):
    response_codes = Counter()

    with open(filename, "r", encoding="utf-8") as f:
        for line in f:
            parts = line.split()

            for i, part in enumerate(parts):
                if part.startswith('HTTP/') and i + 1 < len(parts):
                    code = parts[i + 1]

                    if code.isdigit() and len(code) == 3:
                        response_codes[int(code)] += 1
                    break

    return dict(sorted(response_codes.items()))

how_many_requests("mini-access-log.txt")
```

```python
import os
import json

def print_scores(directory: str):
    files = sorted(os.listdir(directory))
    #print(files)

    for file in files:
        if file.endswith(".json"):
            full_path = os.path.join(directory,file)
            print(full_path)

            with open(full_path,"r",encoding="utf-8") as json_file:
                class_scores = json.load(json_file)
                #print(class_scores)
                scores = {}

                for score in class_scores:
                    for course, result in score.items():
                        scores.setdefault(course,[]).append(result)

                #print(scores)
                for course, results in scores.items():
                    print(f"{course}: min {min(results)}, max {max(results)}, average {sum(results)/len(results):.1f}")


print_scores("scores")
```

```python
# getting file stats in json and extracting insights from that data 

import json
import os
import arrow

def stats_json(filename: str):
    try:
        directory = input("Please type in a directory: ")
        files = os.listdir(directory)
        stats = []

        for file in files:
            full_path = os.path.join(directory,file)
            if not os.path.isfile(full_path):
                continue

            stats.append({
                 "name": file,
                 "size": os.stat(full_path).st_size,
                 "modified": arrow.get(os.stat(full_path).st_mtime).isoformat()
            })

        with open(filename, "w", encoding="utf-8", newline="") as json_file:
                json.dump(stats, json_file, indent=2)

        with open(filename,"r",encoding="utf-8", newline="") as stats_file:
            data = (json.load (stats_file))
            biggest = max(data, key=lambda item: item["size"])
            smallest = min(data, key=lambda item: item["size"])
            newest = max(data, key=lambda item: arrow.get(item["modified"]))
            oldest = min(data, key=lambda item: arrow.get(item["modified"]))

            print("Biggest:", biggest["name"])
            print("Smallest:", smallest["name"])
            print("Newest:", newest["name"])
            print("Oldest:", oldest["name"])
            
    except FileNotFoundError:
        print("That's not a valid directory")
    except PermissionError:
         print("Permission denied accesing")
    
stats_json("output_exercise24_3.json")
```

### **7. Reversing lines**

```python
def reversing_file(reading: str, writing: str):
    with open(reading,"r",encoding="utf-8") as r, \
         open(writing,"w",encoding="utf-8") as w:
        for line in r:
            line = line.rtrip("\n")
            w.write(line[::-1] + "\n")

reversing_file("chapter5_12.txt","output_exercise24.txt")
```

```python
#encrypting and decrypting files

def encrypt_file(input_file: str, encrypted_file: str):
    with open(input_file, "r", encoding="utf-8") as infile, \
        open(encrypted_file, "w", encoding="utf-8") as outfile:
          for char in infile.read():
            outfile.write(str(ord(char)) + " ")

def decrypt_file(encrypted_file: str, decrypted_file: str):
    with open(encrypted_file, "r", encoding="utf-8") as infile, \
        open(decrypted_file, "w", encoding="utf-8") as outfile:
        numbers = infile.read().split()

        for number in numbers:
            outfile.write(chr(int(number)))


encrypt_file("wcfile.txt","encrpyt.txt")
decrypt_file("encrpyt.txt","decrypt.txt")                


    
```

```python
#writing vowels and consonats in two different files


def writing_vowels_consonants(input: str, vowels: str,consonants: str):
    vowels = "aeiou"

    with open(input,"r",encoding="utf-8") as i, \
         open(vowels,"w",encoding="utf-8") as v, \
         open(consonants,"w",encoding="utf-8") as c:


        for line in i:
            line_vowels = ""
            line_consonants = ""

            for character in line.lower():
                if character in vowels:
                    line_vowels += character
                elif character.isalpha():
                    line_consonants += character

            v.write(line_vowels + "\n")
            c.write(line_consonants + "\n")

writing_vowels_consonants("wcfile.txt","vowels.txt","consonants.txt")
```

```python
# creating file with shell and usernames who use the shell 
import csv

def shell(filename: str, output_filename: str):
    shells = {}

    with open(filename, "r", newline="", encoding="utf-8") as passwd:
        read_passwd = csv.reader(passwd, delimiter=":")

        for line in read_passwd:
            if len(line) < 3:
                continue

            username = line[0]
            user_shell = line[-1]

            shells.setdefault(user_shell, []).append(username)

    with open(output_filename, "w", encoding="utf-8") as output:
        for shell_name, users in shells.items():
            output.write(f"{shell_name}:{', '.join(users)}\n")


shell("passwd.txt", "shell.txt")
```
