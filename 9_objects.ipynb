```python
class Book:
    def __init__(self, title, author, price, width):
        self.title = title
        self.author = author
        self.price = price
        self.width = width

class Shelf:
    def __init__(self, width):
        self.books = []
        self.width = width
    
    def add_book(self, *books):
        current_width = sum(book.width for book in self.books)
        new_width = sum(book.width for book in books)

        if current_width + new_width > self.width:
            raise ValueError("Total books width must be less than shelf width")
        
        self.books.extend(books)
        
    
    def __str__(self):
        return ", ".join(book.title for book in self.books)
    
    def clean_shelf(self):
        self.books.clear()

    def total_price(self):
        return sum(book.price for book in self.books)
    
    def has_book(self, title):
        return any(b.title == title for b in self.books)

    
book1 = Book("1984", "George Orwell", 12.50, 4)
book2 = Book("Dune", "Frank Herbert", 18.75, 6)
book3 = Book("The Hobbit", "Tolkien", 15.25, 3)

shelf = Shelf(10)

shelf.add_book(book1, book2)  # 4 + 6 = 10, fits

try:
      shelf.add_book(book3)     # 10 + 3 = 13, doesn't fit
except ValueError as error:
      print(error)


#shelf.clean_shell()
print(shelf)
```

```python
class Envelope:
    def __init__(self, weight: float, was_sent = False):
        self.weight = weight
        self.was_sent = was_sent
        self.current_postage = 0

    def postage_needed(self):
        return self.weight * 10
    
    def add_postage(self, postage):
        self.current_postage += postage

    def send(self):
        if self.current_postage >= self.postage_needed():
            self.was_sent = True

    
class BigEnvelope(Envelope):

    def postage_needed(self):
        return self.weight * 15
    

envelope = BigEnvelope(2.0)
envelope.add_postage(29)
envelope.send()
print(envelope.was_sent)  # False

envelope.add_postage(1)
envelope.send()
print(envelope.was_sent)  # True
```

### **5. Flexible dict**

```python
# Support flexible keys.

class FlexibleDict(dict):
    def __getitem__(self, key):
        try:
            if key in self:
                pass
            elif str(key) in self:
                key = str(key)
            elif int(key) in self:
                key = int(key)
        except ValueError:
            pass

        return dict.__getitem__(self,key)
```

```python
class StringKeyDict(dict):
    def __setitem__(self, key, value):
        key = str(key)
        super().__setitem__(key,value)

skd = StringKeyDict()

skd[1] = 10

print(skd)       # {'1': 10}
print(skd["1"])  # 10
print(skd[1])    # KeyError
```

```python
class RecentDict(dict):
    def __init__(self, max_pairs):
        super().__init__()
        self.max_pairs = max_pairs

    def __setitem__(self, key, value):
        super().__setitem__(key, value)

        if len(self) > self.max_pairs:
            oldest_key = next(iter(self))
            del self[oldest_key]
```

```python
class FlatList(list):

    @staticmethod
    def is_iterable(value):
        try:
            iter(value)
            return True
        except TypeError:
            return False
    
    
    def append(self, value):
        if self.is_iterable(value):
            if value is self:
                value = self.copy()
            for item in value:
                super().append(item)
        else:
            super().append(value)

fl = FlatList()

# Un valor no iterable: se añade normalmente
fl.append(10)
assert fl == [10]

# Una lista: añade sus elementos por separado
fl.append([20, 30, 40])
assert fl == [10, 20, 30, 40]

# Una tupla también es iterable
fl.append((50, 60))
assert fl == [10, 20, 30, 40, 50, 60]

# Una cadena es iterable: añade cada carácter
fl.append("abc")
assert fl == [10, 20, 30, 40, 50, 60, "a", "b", "c"]

# Un iterable vacío no añade nada
fl.append([])
assert fl == [10, 20, 30, 40, 50, 60, "a", "b", "c"]

print("Todas las pruebas pasaron")
  
```

### **8. Zoo**

```python
# Manage cages in a zoo.

class Animal:

    def __init__(self, color):
        self.species = self.__class__.__name__
        self.color = color

    def __repr__(self):
        return f"{self.color} {self.species}, {self.number_of_legs} legs"


class Wolf(Animal):
    number_of_legs = 4
    space_required = 100


class Sheep(Animal):
    number_of_legs = 4
    space_required = 120


class Snake(Animal):
    number_of_legs = 0
    space_required = 22


class Parrot(Animal):
    number_of_legs = 2
    space_required = 34

COMPATIBILITY = {
      Wolf: [Wolf, Sheep],
      Sheep: [Sheep, Wolf],
      Snake: [Snake, Parrot],
      Parrot: [Parrot, Snake],
  }

class Cage:
    space_limit = 350

    def __init__(self, cage_id):
        self.cage_id = cage_id
        self.cage_animals = []

    def add_animals(self, *animals):
        all_animals = self.cage_animals + list(animals)

        occupied_space = sum(
            animal.space_required
            for animal in self.cage_animals
        )

        new_space = sum(
            animal.space_required
            for animal in animals
        )

        if occupied_space + new_space > self.space_limit:
            raise ValueError("Not enough space in the cage")
        
    
        for animal in all_animals:
              compatible_classes = COMPATIBILITY[type(animal)]

              for other_animal in all_animals:
                  if type(other_animal) not in compatible_classes:
                      raise ValueError(
                          f"{type(animal).__name__} is incompatible with "
                          f"{type(other_animal).__name__}"
                      )
        # O(n^2) complexity but it is ok for this exercise
        
        self.cage_animals.extend(animals)



    def __repr__(self):
        animals = "\n".join(str(animal) for animal in self.cage_animals)
        return (
            f"Cage with id {self.cage_id} contains "
            f"{len(self.cage_animals)} animals:\n{animals}"
        )


class BigCage(Cage):
    space_limit = 600

class Zoo:
    def __init__(self):
        self.zoo_cages = []

    def add_cages(self, *cages):
        self.zoo_cages.extend(cages)
    
    def __str__(self):
        return "\n\n".join(repr(cage) for cage in self.zoo_cages)
        
    
    def animals_by_color(self, *colors):
        if not colors:
            raise ValueError("You need to pass at least a color")
        by_color = []
        for cage in self.zoo_cages:
            for animal in cage.cage_animals:
                if animal.color in colors:
                    by_color.append(animal)
        
        return by_color
    
    def animals_by_legs(self, number_of_legs):
        by_legs = []
        for cage in self.zoo_cages:
            for animal in cage.cage_animals:
                if animal.number_of_legs == number_of_legs:
                    by_legs.append(animal)
        return by_legs

    def number_of_legs(self):
        return sum(animal.number_of_legs
                   for cage in self.zoo_cages
                   for animal in cage.cage_animals)
    
    def transfer_animal(self, target_zoo, animal_subclass):
        if self.zoo_cages == [] or self.zoo_cages[0].cage_animals == []:
            print("Still no animals in the zoo")
        else:
            for cage in self.zoo_cages:
                for animal in cage.cage_animals:
                    if type(animal) == animal_subclass:
                        cage.cage_animals.remove(animal)
                        target_zoo.zoo_cages[0].cage_animals.append(animal)

    def get_animals(self, **kwargs):
        valid_keys = {"color","legs"}

        if not kwargs:
            raise ValueError("You need to pass color and/or legs")
        if set(kwargs) - valid_keys:
            raise ValueError("valid filters are only color and legs")
        
        result = []

        for cage in self.zoo_cages:
            for animal in cage.cage_animals:
                if "color" in kwargs and animal.color != kwargs["color"]:
                    continue

                if "legs" in kwargs and animal.number_of_legs != kwargs["legs"]:
                    continue

                result.append(animal)

        return result
                        


    
# wolf = Wolf("black")
# sheep = Sheep("white")
# snake = Snake("green")
# parrot = Parrot("white")

# c1 = Cage(1)
# c1.add_animals(wolf, sheep)

# c2 = Cage(2)
# c2.add_animals(snake, parrot)

# z = Zoo()
# z.add_cages(c1, c2)

# print(z)
# print(z.animals_by_color("white"))
# print(z.animals_by_color("black"))
# print(z.animals_by_color("purple"))
# print(z.animals_by_legs(4))
# print(z.number_of_legs())
# print(z.animals_by_color("white", "black"))
wolf = Wolf("black")
sheep = Sheep("white")
snake = Snake("green")
parrot = Parrot("blue")

source_cage = Cage(1)
source_cage.add_animals(wolf, sheep)

target_cage = Cage(2)
target_cage.add_animals(Wolf("gray"))

source_zoo = Zoo()
source_zoo.add_cages(source_cage)

target_zoo = Zoo()
target_zoo.add_cages(target_cage)

source_zoo.transfer_animal(target_zoo, Wolf)

print("SOURCE ZOO")
print(source_zoo)

print("\nTARGET ZOO")
print(target_zoo)

source_zoo.get_animals(color="white")
```
