### **1. MyEnumerate**

```python
# Recreate enumerate manually.

class MyEnumerate:
    def __init__(self, iterable):
        self.index = 0
        self.iterable = iterable

    def __iter__(self):
        return self
    
    def __next__(self):
        if self.index >= len(self.iterable):
            raise StopIteration

        value = (self.index, self.iterable[self.index])
        self.index += 1

        return value  


for indice, letra in MyEnumerate("abc"):
    print(f"{indice}: {letra}")

nombres = ["Ana", "Luis", "Marta"]
for indice, nombre in MyEnumerate(nombres):
    print(indice, nombre)

resultado = list(MyEnumerate([10, 20, 30]))
print(resultado)

list(MyEnumerate(42))

# if we had a non indexable iterable we can use the following code:

# class MyEnumerate:
#       def __init__(self, data):
#           self.iterator = iter(data)  # Lanza TypeError si data no es iterable
#           self.index = 0

#       def __iter__(self):
#           return self

    # def __next__(self):
    #         value = next(self.iterator)  # Lanza StopIteration al terminar

    #         result = (self.index, value)
    #         self.index += 1

    #         return result
```

### **2. Circle**

```python
# Cycle values repeatedly.

class Circle:
    def __init__(self, iterable, number):
        self.number = number
        self.iterable = iterable

    def __iter__(self):
        return CircleIterator(self.iterable, self.number)
    
class CircleIterator:
    def __init__(self, iterable, number):
        self.iterable = iterable
        self.number = number
        self.index = 0
        self.counter = 0

    def __iter__(self):
        return self
    
    def __next__(self):
        if self.index >= len(self.iterable):
            self.index = 0

        if self.counter >= self.number or not self.iterable:
            raise StopIteration
        
        value = self.iterable[self.index]
        self.index += 1
        self.counter += 1

        return value
    

c = Circle('abc', 5)
print(list(c)) 

# cuter version
# class Circle():
# def __init__(self, data, max_times):
# self.data = data
# self.max_times = max_times
# def __iter__(self):
# n = len(self.data)
# return (self.data[x % n] for x in range(self.max_times))
```

```python
class CircleIterator:
    def __init__(self, iterable, number):
        self.iterable = iterable
        self.number = number
        self.index = 0
        self.counter = 0

    def __iter__(self):
        return self
    
    def __next__(self):
        sequence = getattr(self, self.returns[0])

        if self.index >= len(sequence):
            self.index = 0

        if self.counter >= self.number or not sequence:
            raise StopIteration
        
        value = sequence[self.index]
        self.index += 1
        self.counter += 1

        return value
    
class Circle(CircleIterator):
    def __init__(self, iterable, number):
        self.returns = ["iterable"]
        super().__init__(iterable, number)
    
    
    
c = Circle('abc', 5)
print(list(c))
```

```python
def circle(iterable, number):
    if not iterable:
        raise ValueError("Sequence mustn't be empty")
    
    for index in range(number):
        yield iterable[index % len(iterable)]


c = circle('abc', 5)
print(list(c))
```

### **3. All lines, all files**

```python
# Iterate lines across files.

import os

def all_lines_all_files(directory: str):
    for filename in os.listdir(directory):
        path = os.path.join(directory, filename)

        try:
            with open(path, "r", encoding="utf-8") as f:
                for line in f:
                    yield line
        except OSError:
            pass

file = all_lines_all_files(r"C:\Users\Usuario\Desktop\PROGRAMMING\GENERAL PYTHON\python_workout_50_exercises\books")
print(next(file))
print(next(file))
```

### **4. Elapsed since**

```python
# Yield elapsed times.

import time


def elapsed_since(data):
    last_time = None

    for item in data:
        current_time = time.perf_counter()

        delta = current_time - (last_time or current_time)

        last_time = time.perf_counter()

        yield (delta, item)

g = elapsed_since(["a", "b", "c"])

print(next(g))      # (0, "a")

time.sleep(2)
print(next(g))      # algo como (2, "b")

time.sleep(1)
print(next(g))      # (1, "c")
```

```python
import time

def elapsed_since_extended(iterable, t):
    last_time = None

    for item in iterable:
        current_time = time.perf_counter()
        
        if last_time is not None:
            delta = current_time - last_time
            if delta < t:
                time.sleep(t - delta)
                delta = t
        else:
            delta = 0

        last_time = time.perf_counter()

        yield delta, item



g = elapsed_since_extended("abc", 2)

start = time.time()

print(next(g))  # esperado: "a", inmediatamente
print(round(time.time() - start))  # esperado: 0

print(next(g))  # esperado: espera hasta que hayan pasado 2s desde el anterior
print(round(time.time() - start))  # esperado: 2

print(next(g))  # esperado: espera otros 2s
print(round(time.time() - start))  # esperado: 4
```

```python
import os
from datetime import datetime

def file_usage_timing(directory: str):
    for filename in  os.listdir(directory):
        path = os.path.join(directory, filename)

        try:
            stats = os.stat(path)

            atime = datetime.fromtimestamp(stats.st_atime)
            mtime = datetime.fromtimestamp(stats.st_mtime)
            ctime = datetime.fromtimestamp(stats.st_ctime)
            yield(filename, atime.strftime('%d/%m/%Y %H:%M:%S'), mtime.strftime('%d/%m/%Y %H:%M:%S'), ctime.strftime('%d/%m/%Y %H:%M:%S'))
        except OSError:
            print("A problem was found")

f = file_usage_timing(r"C:\Users\Usuario\Desktop\PROGRAMMING\GENERAL PYTHON\python_workout_50_exercises\books")
print(next(f))
print(next(f))
```

### **5. MyChain**

```python
# Chain iterables together.

def my_chain(*iterables):
    for iter in iterables:
        for item in iter:
            yield item

for one_item in my_chain('abc', [1,2,3], {'a':1, 'b':2}):
    print(one_item)

```

```python
def my_zip(*iterables):

    iterators = [iter(iterable) for iterable in iterables]

    while True:
        reached_shortest = False
        result = []

        for iterable in iterators:
            try:
                result.append(next(iterable))
            except StopIteration:
                reached_shortest = True
                break

        if reached_shortest:
            break
        yield tuple(result)

print(list(my_zip("abc", [10, 20, 30])))
print(list(my_zip([1, 2, 3], ["a", "b"])))
print(list(my_zip([0, False, None], ["x", "y", "z"])))
```

```python
_MISSING = object()

def my_range_generator_expression(start, final, step=1):
    if start is _MISSING and final is _MISSING:
        raise TypeError("Myrange needs at least an argument")

    if final is _MISSING:
        final = start
        start = 0
    elif start is _MISSING:
        start = 0

    if step == 0:
        raise ValueError("step must not be zero")

    start = start
    final = final
    step = step
    current = start

    if step > 0:
        while current < final:
            yield current
            current += step
    else:
        while current > final:
            yield current
            current += step

for i in my_range_generator_expression(0,10,2):
    print(i)

for i in my_range_generator_expression(20,0,-2):
    print(i)
```

```python
from itertools import count, takewhile

_MISSING = object()

def my_range(start=_MISSING, final=_MISSING, step=1):
    if start is _MISSING and final is _MISSING:
        raise TypeError("my_range needs at least one argument")

    if final is _MISSING:
        final = start
        start = 0

    if step == 0:
        raise ValueError("step must not be zero")

    if step > 0:
        return (x for x in takewhile(lambda x: x < final, count(start, step)))

    return (x for x in takewhile(lambda x: x > final, count(start, step)))
```
